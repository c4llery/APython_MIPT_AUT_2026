```python
# Задание 1

def mult():
    for i in range(1, 10):
        for j in range(1, 10):
            print(f'{j} * {i} = {i * j:^3}', end = '  ')
        print()

def plus():
    for i in range(1, 10):
        for j in range(1, 10):
            print(f'{j} + {i} = {i + j:^3}', end = '  ')
        print()
        
def minus():
    for i in range(1, 10):
        for j in range(1, 10):
            print(f'{j} - {i} = {j - i:^3}', end = '  ')
        print()
def div():
    for i in range(1, 10):
        for j in range(1, 10):
            print(f'{j} / {i} = {j / i:^10.5}', end = '  ')
        print()
        
print('Выберите, какую табличку хотите посмотреть:')
print('+ - сложение')
print('- - вычитание')
print('/ - деление')
print('* - умножение')
symbol = input()
if (symbol == '+'):
    plus()
elif (symbol == '-'):
    minus()
elif (symbol == '*'):
    mult()
elif (symbol == '/'):
    div()
else:
    print('Ошибка!')
```
