```python
# Задание 1

def mult():
    for i in range(1, 10):
        for j in range(1, 10):
            print(f'{j} * {i} = {i * j:^3}', end = '  ')
        print()

def plus():
    for i in range(1, 10):
        for j in range(1, 10):
            print(f'{j} + {i} = {i + j:^3}', end = '  ')
        print()
        
def minus():
    for i in range(1, 10):
        for j in range(1, 10):
            print(f'{j} - {i} = {j - i:^3}', end = '  ')
        print()
def div():
    for i in range(1, 10):
        for j in range(1, 10):
            print(f'{j} / {i} = {j / i:^10.5}', end = '  ')
        print()
        
print('Выберите, какую табличку хотите посмотреть:')
print('+ - сложение')
print('- - вычитание')
print('/ - деление')
print('* - умножение')
symbol = input()
if (symbol == '+'):
    plus()
elif (symbol == '-'):
    minus()
elif (symbol == '*'):
    mult()
elif (symbol == '/'):
    div()
else:
    print('Ошибка!')
```

```python
# b
def create_matrix_by_size(M, N):
    m = [[random.random() for i in range(N)] for j in range(M)]
    print(m)
   
create_matrix_by_size(3, 2)
```

```python
# c 
def matrix_vector_multiply(matr, vec):
    res = []
    for i in range(len(matr)):
        ans = 0
        for j in range(len(matr[i])):
            ans += matr[i][j] * vec[j] 
        res.append(ans)
    return res

matrix_vector_multiply([[1, 2, 3], [4, 5, 6]], [1, 1, 1])
```

```python
# d
def print_matrix(matr):
    for i in range(len(matr)):
        for j in range(len(matr[i])):
            print(f'{matr[i][j]:^5}', end = '')
        print()
print_matrix([[1, 2, 3], [4, 5, 6]])
    
```

```python
# e 
def print_vector(vec):
    for i in range(len(vec)):
        print(vec[i], end = ' ')
        
print_vector([1, 2, 3])
```

```python
# f
def find_sum_diag_elems(matr):
    print_matrix(matr)
    sum = 0
    N = len(matr[0])
    for i in range(N):
        sum += matr[i][i]
    
    return(sum)
find_sum_diag_elems([[1, 2, 3], [4, 5, 6], [7, 8, 9]])
```

```python
# g
def find_conv(img, kern):
    img_x = len(img[0])
    img_y = len(img)
    kern_x = len(kern[0])
    kern_y = len(kern)
    ans = [[0 for i in range(img_x - kern_x + 1)] for j in range(img_y - kern_y + 1)]
    for i in range(img_y - kern_y + 1):
        for j in range(img_x - kern_x + 1):
            res = 0
            for k in range(kern_y):
                for l in range(kern_x):
                    res += kern[k][l] * img[i+k][j+l]
            ans[i][j] = res
    return ans

find_conv([[1, 2, 3], [4, 5, 6], [7, 8, 9]], [[1, 0], [0, 1]])
```

```python
def multichannel_processing(func):
    def wrapper(data, kernel):
        ans = [0 for i in range(3)] 
        for i in range(3):
            ans[i] = func(data[i], kernel)
        return ans
    return wrapper
    
    
@multichannel_processing
def find_conv_mult(img, kern):
    img_x = len(img[0])
    img_y = len(img)
    kern_x = len(kern[0])
    kern_y = len(kern)
    ans = [[0 for i in range(img_x - kern_x + 1)] for j in range(img_y - kern_y + 1)]
    for i in range(img_y - kern_y + 1):
        for j in range(img_x - kern_x + 1):
            res = 0
            for k in range(kern_y):
                for l in range(kern_x):
                    res += kern[k][l] * img[i+k][j+l]
            ans[i][j] = res
    return ans

multi_image = [
    [                       
        [1, 2, 3],
        [4, 5, 6],
        [7, 8, 9],
    ],
    [                       
        [1, 2, 3],
        [4, 5, 6],
        [7, 8, 9],
    ],
    [                       
        [1, 2, 3],
        [4, 5, 6],
        [7, 8, 9],
    ],
]

find_conv_mult(multi_image, [[1, 0], [0, 1]])
```
