```python
import datetime

class LogInfo:
    def __init__(self):
        self.date = 0
        self.level = 0
        self.source = 0
        self.message = ""
        self.id = 0
        self.user = 0
        self.extra_info = ""

    def create_string(self):
        info = ""
        info += "==========\n"
        info += f"Дата: {self.date}\n"
        info += f"Уровень серьёзности: {self.level}\n"
        info += f"Источник события: {self.source}\n"
        info += f"Сообщение: {self.message}\n"
        info += f"Идентификатор события: {self.id}\n"
        info += f"Пользователь/сессия: {self.user}\n"
        info += f"Дополнительная информация: {self.extra_info}\n"
        info += "==========\n"
        return info

 
class Logger:
    def __init__(self, file_path):
       assert isinstance(file_path, str) 
       self.path = file_path

    def create_log(self, level, source, message, event_id = 0, user = "", extra_info = ""):
        loginfo = LogInfo()
        loginfo.date = datetime.datetime.now().strftime("%Y-%m-%d %H:%M:%S")
        loginfo.level = level
        loginfo.source = source
        loginfo.message = message 
        loginfo.id = event_id
        loginfo.user = user
        loginfo.extra_info = extra_info
        with open(self.path, "a", encoding = "utf-8") as f:
            f.write(loginfo.create_string())


   

```

```python
logger = Logger("Log.txt")
logger.create_log("INFO", "AuthModule", "Пользователь вошёл в систему",
            event_id=11, user="ivanov")
```
