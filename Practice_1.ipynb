```python
# Задача 2
k = 1
for i in range(5):
    if (i % 2 == 0):
        print('* {} * {} *'.format(k, k+1))
        k += 2
    else:
        print('{} * {} * {}'.format(k, k+1,k+2))
        k += 3
```
