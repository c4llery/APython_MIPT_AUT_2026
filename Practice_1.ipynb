```python
# Задача 3

print('Поиск числа')
print('Для ответов вводите Да/Нет')
print('Введите интервал, где расположено ваше число')
a, b = map(int, input().split())
a, b = min(a, b), max(a, b)
N = 0
end = 0
while (end != 1):
    if (b - a <= 1):
        print(f'Число равно {a}?')
        ans = input()
        if (ans == 'Да'):
            N = a
            end = 1
            print(f'Ваше число: {N}')
        elif (ans == 'Нет'):
            print(f'Число равно {b}?')
            ans = input()
            if (ans == 'Да'):
                N = b
                end = 1
        else:
            print('Ошибка')
            break
    else:
        N = (b + a) // 2
        print(f'Число равно {N}?')
        ans = input()
        if (ans == 'Да'):
            end = 1
            print(f'Ваше число: {N}')
        elif (ans == 'Нет'):
            print(f'Число меньше {N}?')
            ans = input()
            if (ans == 'Да'):
                b = N
            elif (ans == 'Нет'):
                a = N
            else:
                print('Ошибка')
                break
        else:
            print('Ошибка')
            break
```

```python
# Задача 4

print('Вводите числа больше 0')
max_num = -1
num = 100
while (num > 0):
    num = int(input())
    if (num < 0):
        print('Ошибка!')
        max_num = -1
        break
    max_num = max(num, max_num)
if (max_num > 0):
    print(f'Максимальное число: {max_num}')
```
